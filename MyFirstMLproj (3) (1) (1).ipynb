```python
def fix_outliers_iqr(df):
    num=df.select_dtypes(include='number').columns
    for col in num:
        Q1=df[col].quantile(0.25)
        Q3=df[col].quantile(0.75)
        IQR= Q3 - Q1
        lower_bound=  Q1 - 1.5 * IQR
        upper_bound= Q3 + 1.5 * IQR
        df[col]=df[col].apply(lambda x: lower_bound if x < lower_bound else upper_bound if x > upper_bound else x)
    return df    
    
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

ohe = OneHotEncoder(sparse_output=False, drop=None)

# 1. First column
caec = ohe.fit_transform(df1[['food_between_meals_freq']])
cols = ohe.get_feature_names_out(['food_between_meals_freq'])
df_encoded = pd.DataFrame(caec, columns=cols, index=df1.index) 
merge = pd.concat([df1.drop('food_between_meals_freq', axis=1), df_encoded], axis=1)

# 2. Second column
calc = ohe.fit_transform(merge[['alcohol_consumption_freq']])
cols = ohe.get_feature_names_out(['alcohol_consumption_freq'])
df_encode = pd.DataFrame(calc, columns=cols, index=merge.index) 
merge2 = pd.concat([merge.drop('alcohol_consumption_freq', axis=1), df_encode], axis=1)

# 3. Third column
mtrans = ohe.fit_transform(merge2[['transportation_type']])
cols = ohe.get_feature_names_out(['transportation_type'])
df_encode1 = pd.DataFrame(mtrans, columns=cols, index=merge2.index) 

# SAFE: Save to df_clean so df1 is never ruined if you rerun this cell!
df_clean = pd.concat([merge2.drop('transportation_type', axis=1), df_encode1], axis=1)

print(df_clean)
```

```python
# Step 7: SMOTE

from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

# Separate X and y
X = df_clean.drop('target', axis=1)
y = df_clean['target']

# Apply SMOTE
x_smote, y_smote = smote.fit_resample(X, y)

# Create balanced DataFrame
balanced_df = pd.DataFrame(x_smote, columns=X.columns)

# Add target column
balanced_df['target'] = y_smote

# Check the result
print("Original shape:", df_clean.shape)
print("Balanced shape:", balanced_df.shape)

print("\nTarget distribution after SMOTE:")
print(balanced_df['target'].value_counts())

balanced_df.head()
```

```python
#Store results
scores = []
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy= accuracy_score(y_test,y_pred)
    precision= precision_score(y_test,y_pred,average='weighted')
    f1= f1_score(y_test,y_pred,average='weighted')
    recall= recall_score(y_test,y_pred,average='weighted')
    scores.append({
        "Model": name,
        "accuracy": round(accuracy, 2),
        "precision": round(precision, 2),
        "f1": round(f1, 2),
        "recall": round(recall, 2)
    })
# Convert to DataFrame for nice display
results_df = pd.DataFrame(scores)
print(results_df)
```
